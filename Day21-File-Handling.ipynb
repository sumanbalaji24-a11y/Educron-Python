# File Handling

## Open(filename, mode)

## Read('filename','a')
1. Just reads the data from the provided file
2. If file doesn't exist throws file not foud error

```python
try:
    with open('demo.txt','r') as f:
        context = f.read()
        print(context)
except Exception as exc:
    print(exc)
```

```python
print(f)
print(context)
```

## write('filename','w') 
  1. creates a new file and writes into it
  2. If file already exists overwrites the data in it 

```python
try:
    with open('demo1.txt','w') as f:
      f.write("hello world  how are you  what is going on man\n")
      f.write("hello world  how are you what is going on boy")
     
except Exception as exc:
    print(exc)
```

## Append('filename','a')
1. Opens the file and moves the cursor to the end of the existing data

```python
try:
    with open('demo1.txt','a') as f:
         f.write("how are you doing in life\n")
         f.write("how are you doing in life\n")
except Exception as exc:
    print(exc)
```

## Create('filename','x')
1. Creates a file if not exist
2. If exists it throws an error

```python
try:
    with open('demo2.txt','x') as f:
         f.write("how are you doing in life\n")
         f.write("how are you doing in life\n")
except Exception as exc:
    print(exc)
```

## Read+Write open('filename','r+')
1. cursor starts at the begining and the file must exist

```python
try:
    with open('demo2.txt','r+') as f:
        oldcontent = f.read()
        print(f"old: {oldcontent}")
        f.write("\n good morning")
        newcontent = f.read()
        print(f"new: {newcontent}")
except Exception as exc:
    print(exc)
```
