## Range
Range(start,stop,step)
start - inclusive
stop- exclusive
step - how many steps

```python
## Range (stop) - starts from 0 to stop
r = range(10)
print(r)
for  i in r:
    print(i)
print(f"length of range is {len(r)}")
```

```python
## Range(start,stop) - starts from start and ends at stop
for i in range(10,15):
    print(i)
r = range(1,10)
listItem = list(r)  ## we can view the elements in the range
print(listItem)
tupleItem = tuple(r)  ## we can view the elements in the range
print(tupleItem)
setItems = set(r)
print(setItems)

```

```python
# multiplication of 5
for i in range(1,11):
    print(f" 5 x {i} = {5*i}")
```

```python
## Input
number = int(input("Enter number"))
for i in range(1,11):
    print(f" {number} x {i} = {number * i}")
```
