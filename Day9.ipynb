# Set union(|) Insertion(&) Difference(-) Semantic Difference(^)
## Union
  Returns all the elemenst from both the sets and removes duplicates
## Insertion 
  Returns matching elements from both the sets
## Difference
   returns all from set a and matching from set b
## Semantic Difference
    Returns non matching elements from both the sets


```python
## union (|)
numbers1 = {1,2,3,4}
numbers2 = {4,5,5,6,7,8}
res = numbers1 | numbers2
res
```

```python
## intersection
numbers1 = {1,2,3,4}
numbers2 = {4,5,6,7,8}
res = numbers1 & numbers2
res
```

```python
## Difference returns items present in first set not in second
numbers1 = {1,2,3,4}
numbers2 = {4,5,6,7,8}
res = numbers1 - numbers2
print(res)
res = numbers2 - numbers1
print(res)
```

```python
# Semantic Difference  returns non matching elements from both sets (opp of intersection)
numbers1 = {1,2,3,4}
numbers2 = {4,5,6,7,8}
res = numbers1 ^ numbers2
print(res)
```

## Subset and super set 

```python
## Subset
set_1 = {1,2,3}
set_2 = {1,2,3,4,5}
print(set_1.issubset(set_2))
print(set_2.issubset(set_1))
```

```python
## Super set 
set_1 = {1,2,3}
set_2 = {1,2,3,4,5}
print(set_1.issuperset(set_2))
print(set_2.issuperset(set_1))
```
