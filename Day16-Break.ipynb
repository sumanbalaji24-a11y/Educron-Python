# break statement

```python
for i in range(1,10):
    print("start loop")
    print(i)
    if(i==5):
        continue
    print("End loop")
```

```python
students = ["anil",'amit','balaji']
search = 'amit'
for student in students:
    print(student)
    if(student.lower == search.lower):
        print(f"found search {student}")
        break
```

## Continue
1. when ever the condition is found skip the code below and move to the next iteration


```python
for i in range(1,10):
    if(i % 2 == 0):
        continue
    print(i)   ## gets skipped when i is even
```

```python
## print only possitive numbers
numbers = [1,-2,-3,4,5]
for i in numbers:
    if(i<0):
        continue
    print(i)
```

```python
for i in range(1,5):
    print(i)
    for j in range(1,5):
        print(j)
        if(j==2):
            break
```
