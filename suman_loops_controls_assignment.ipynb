Rules:
- A year is a leap year if it is divisible by 4
- However, if it is divisible by 100, it is NOT a leap year
- Exception: If it is also divisible by 400, then it IS a leap year

Examples:
- 2000: Leap year (divisible by 400)
- 1900: Not a leap year (divisible by 100 but not 400)
- 2024: Leap year (divisible by 4 but not 100)

**Hint:** Check conditions in order: first divisible by 400, then by 100, then by 4. Use modulo operator (%) to check divisibility.

````python
def check_leap_year(year):
    if year < 0:
        return " invalid year entered. please enter the valid year"
    if(year % 400 ==0):
        return "leap year"
    elif(year%100 ==0):
        return "not a leap year"
    elif(year%4 ==0):
        return "leap year"
      
print(check_leap_year(2000))
print(check_leap_year(1900))
print(check_leap_year(2024))
````

### Problem 6.7: Pascal's Triangle (Hard)


````python
n =5
prevRow = []
for i in range(0,n):
    currRow =[]
    for j in range(0,i+1):
        if(j ==0 or j==i):
            currRow.append(1)
        else:
            currRow.append(prevRow[j-1]+prevRow[j])
    print(f"{' ' * (n - i - 1)} {' '.join(map(str, currRow))}")
    prevRow = currRow
````
