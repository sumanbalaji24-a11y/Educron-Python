# Dictionary and range

```python
## check if the key exist
person ={"name":"suman","age":20}
print(person.keys()) ## returns keys
print(person.values()) ## returns values
print(person.items()) ## returns all values as tuples

## check key exists or not using input
print("name" in person)
print("address" in person)
print("address" not in person)

## check values exists in person.Values
print(person)
print(20 in person.values())
print("suman" in person.values())
```

## Nested Dictionary

```python
users= {
    "user1":{
        "name" :"suman",
        "age" : 20,
        "city" :"Banagalore"
    },
    "user2":{
            "name" :"kiran",
            "age" : 30,
            "city" :"Banagalore"
    },
    "user3":{
                "name" :"srinivasa",
                "age" : 49,
                "city" :"Kolar"
        }
}
## accessing nested elements
print(users.get("user2").get("age"))
print(users["user2"].get("brand","unknown"))
```
