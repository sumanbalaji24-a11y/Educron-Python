## Boolean
1. Represents one of two truth always: True or False
2. True Corresponds to 1
3. False Corressponds to 0
4. Comparision Operators and logical operators awalys gives us a boolean value
5. and and or operator
6. not 

```python
isActive = True
isVerified = False
print(isActive and isVerified)
print(isVerified or isActive)
print(not not isActive)
```
