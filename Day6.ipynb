## Slicing

```python
## Slicing
message2 = "hello"
print(message2[::])
print(message2[::1])
print(message2[::2])
print(message2[::-1])

```

## Join

```python
messages = ["hello", "world", "how", "are", "you"]
print(", ".join(messages))
```

## StartsWith and EndsWith (Case Sensitive, Start, End)

```python
message3 = 'hello'
print(message3.startswith('he')) # Case Sensitive
print(message3.startswith('He')) # Case Sensitive
print(message3.startswith('hello')) # Case Sensitive
print(message3.endswith("llo")) 
print(message3.endswith("hello")) 
print(message3.endswith("llpoppo")) 
message4 = "suman.pdf"
print(message4.endswith('.pdf'))
print(message4.startswith(".pdf",5))
print(message4.endswith(".pdf",5,9))
```
