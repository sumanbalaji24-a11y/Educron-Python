## Inbuilt functions

```python
print(min([1,2,3,4]))
print(max([1,2,3,40]))

print(sum([1,2,3,4]))
print(sum([1,2],2))
```

## User Defined Function

```python
def calculate_avg(group):
    total = sum(group)
    avg = total/len(group)
    return avg
group1 = [12,13,15]
calculate_avg(group1)
```
