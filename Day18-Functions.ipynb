# User Defined Functions

```python
## Func With Parameters and returning a result
def add(a,b):
    result = a+b
    return result
```

```python
add(1,2)
```

## Paramater vs Args

```python
## here the parameters are name,age,gender
def student_info(name,age,gender):
    print(name)
    print(age)
    print(gender)
```

```python
student_info("suman",12,"M") ## These values are arguments 
```

## Passing args with parameter names


```python
## even though the order is changed the method matches them accordingly when you pass the actual parameter name
def student_info(name,age,gender):
    print(f"Name:{name}")
    print(f"Age:{age}")
    print(f"Gender:{gender}")
```

```python
## even though the order is changed the method matches them accordingly when you pass the actual parameter name
student_info(age=10,name="suman",gender='M') 
student_info(gender='M',name="test",age=20) 
```

## Default Value For Parameters

```python
def student_info(name,age,gender="M"):
    print(f"Name:{name}")
    print(f"Age:{age}")
    print(f"Gender:{gender}")
```

```python
student_info("suman",12) ## Gender will M by default even when you are not passing it
student_info("shyla",12,'F') ## takes passed value when provided
```

## Variable Len Args (*Args) 
Input will be tuple

```python
def print_args(*args):
    print(f"Args Type is: {type(args)}")
    print(f"Values:{args}")
```

```python
print_args(1,2,3,4)
```

```python
# Note: *numbers is a tuple here 
def sum_numbers(*numbers):
    """ this func returns sum of numbers"""
    total = 0
    for num in numbers:
        total+= num;
    return total
```

```python
sum_numbers(1,2,3,4,5)
```

```python
help(sum_numbers)
```

## Variable Length Keyword Args (**KwArgs)
1. Input will be dict here

```python
student = {"id":1, "name":"suman"}
for key,value in student.items():
    print(f"{key}:{value}")
```

## Combine All Arg types
Order Matters : positional, *args, default, **kwargs

```python
def ex_fn(a,b,*args,c=10,**kwargs):
    print(f"a:{a}")
    print(f"b:{b}")
    print(f"args:{args}")
    print(f"c:{c}")
    print(f"kwargs:{kwargs}")

```

```python
ex_fn(10, 20, 1, 2, 3, 4, 5, id=1, name="suman")
```

## Return Single vs Multiple Vaues
Single returns single value
mutiple returns will return a tuple

```python
## Returns a single value
def get_full_name(firstName,lastName):
    fullName = firstName + lastName;
    return fullName;
```

```python
print(get_full_name("suman","balaji"))
```

```python
## Returns Multiple Values has a tuple
def get_full_name(firstName,lastName):
    return firstName, lastName;
```

```python
res = get_full_name("suman","balaji")
print(res) ## res here is a print
```
