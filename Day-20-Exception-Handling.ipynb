# Excception Handling

Exception is an error thaat occurs during program execution in python.
When an exception occurs python throws an error and stops the execution.

```python
name = "test" 
def print_name():
    name += "Hello"
    print(name)
```

## Generic Exception

```python
print("start")
try:
    print_name()
except Exception as e :
    print(f"Error:{e}")
print("end")
```

## Catch Specific Exception

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Error: Can't divide a number by 0") 
```

## Catch Mutiple Exceptions

```python
try:
    user_input = int(input("Enter a number"))
    result = 10 / user_input
except ValueError:
    print("enter a valid digit")
except ZeroDivisionError:
    print("Error: enter a valid number")
```

## Finally
Ir-respective of whether try or finally gets executed finally will be always executed

```python
try:
    user_input = int(input("Enter a number"))
    result = 10 / user_input
    print(result)
except ValueError:
    print("enter a valid digit")
except ZeroDivisionError:
    print("Error: enter a valid number")
finally:
    print("Finally always gets executed")
```

## Else 
Gets invoked only when try is sucessfull

```python
try:
    user_input = int(input("Enter a number"))
    result = 10 / user_input
    print(result)
except ValueError:
    print("enter a valid digit")
except ZeroDivisionError:
    print("Error: enter a valid number")
else:
    print("Works only when try is sucessfully executed")
finally:
    print("Finally always gets executed")
```
