```python
# reverse a string
name = "suman12"
result = ""
for i in range(len(name),0,-1):
    result+=name[i-1]
print(result)
```

```python
# check string is palindrome or not
user_input = input("enter a string")
result = ""
for char in user_input:
    result = char + result;
if(user_input == result):
    print("it's a palindrome")
else:
    print("not a palindrome")
```

```python
numbers = [1,2,3,4,15,3,4,3,1,2]
unique_elements = list(set(numbers))
repeated = []
for i in unique_elements:
    if(len(list(filter(lambda x: x==i, numbers))) > 1):
        repeated.append(i)
print(repeated)      
```

```python
numbers = [1,2,3,4,15,3,4,3,1,2]
seen = set()
duplicate_elements = []

for i in numbers:
    if i in seen and i not in duplicate_elements:
        duplicate_elements.append(i)
    else:
        seen.add(i)
    
print(duplicate_elements)
```

```python
s = "We are learning python . Python is fun for learning learning"
# print(s.lower().replace('.',""))
repeated = {}

for i in s.lower().replace('.',"").split():
    if i in repeated.keys():
        repeated[i] = repeated[i]+1
    else:
        repeated[i] = 1
print(repeated)
```

```python
# 5. Find the **second-largest number** in a list.
numbers = [1,23,45,3,67]
largest = 0
secondLargest = 0
for i in numbers:
    if(i>largest):
        secondLargest = largest
        largest = i
    elif(i>secondLargest and secondLargest !=i):
        secondLargest = i
print(secondLargest)
```

```python
# 5. Find the **second-lowest number** in a list.
numbers = [1,2,23,45,3,67]
lowest = float('inf')
secondLowest = float('inf')
## low=1, slow=inf
for i in numbers:
    print(i)
    if(i<lowest):
        secondLowest = lowest
        lowest = i
    elif(i > lowest and i < secondLowest):
        secondLowest = i
print(secondLowest)
```

```python
# 7. Find the first non-repeating character.
numbers = [1,3,1,1,4,5,1]
result = {}
firstNonRepeatedNumber = 0
for i in numbers:
    if(i in result):
      result[i]+=1
    else:
        result[i]=1
for key,value in result.items():
   if(value==1):
      firstNonRepeatedNumber = key
      break
print(firstNonRepeatedNumber)      

   
```

```python
# 9. Find missing number from `1` to `n`.
numbers = [3,4,2,5]
n = len(numbers)+1
expectedSum = (n*(n+1))//2
actualSum = 0
for i in numbers:
    actualSum+=i
print(expectedSum - actualSum)
```
