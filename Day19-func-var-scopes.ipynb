# Local And Global Variables

```python
name = 'suman'
def print_name():
    name = "suman balaji"
    print(name)
    x= "suman"
    print(id(name))
    print(id(x))
```

```python
print(id(name))
print_name()
```

## You can't modify global variable locally without using global keyword

```python
count = 0
def increment():
    count+=1   ## throws UnboundLocalError: cannot access local variable 'count' where it is not associated with a value
    print(count)
```

```python
increment()
```

```python
count = 0
def increment1():
    global count
    count+= 1
    print(count)
```

```python
increment1()
increment1()
```
