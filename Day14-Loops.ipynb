## Enumerate

```python
elements =  [20,10,4,10,23]
for index,item in enumerate(elements):
    print(f"{index}: {item}")

elements1 =  [20,10,4,10,23]
for index,item in enumerate(elements1,start=10):
    print(f"{index}: {item}")
```

```python
names=["a","b","c"]
ages = [12,13,14]

for i,item in enumerate(names):
    print(f"{item}:{ages[i]}")
```

## Zip

```python
names=["a","b","c"]
ages = [12,13,14]

for name,age in zip(names,ages):
    print(name,age)
```
