# Lamda Function
Syntax: Lamda arguments: Expression

```python
add_numbers = lambda a,b : a+b
```

```python
add_numbers(1,2)
```

```python
square_numbers = lambda a : a**a
```

```python
square_numbers(3)
```

```python
area_of_circle = lambda length,width : length * width
```

```python
area_of_circle(10,20)
```

```python
is_even = lambda num:True if num % 2 == 0 else False
```

```python
is_even(10)
```

```python
is_even(13)
```

## Map with lambda
Syntax: Map(function,iterable)

```python
numbers = [1,2,3,4]
result = [i*i for i in numbers]
```

```python
result
```

```python
numbers = [1,2,3,4]
multiply_self = lambda a : a*a
result = map( lambda a : multiply_self(a) , numbers)
print(list(result))
```

## Filter with Lambda
Syntax - filter(function, iterable)

```python
numbers = [1,2,3,4]
multiply_self = lambda a : a*a
result = filter( lambda a : a % 2!=0 , numbers)
print(list(result))
```

```python
words = ["apple","cat","dog"]
result = list(filter(lambda x: len(x) ==3, words))
print(result)
```

## Sorted With Lambda
Syntax : sorted(iterable, key=lambda argument : expression, reverse=True/False)

```python
numbers = [49,14,3,2]
result = sorted(numbers)
print(result)
```

```python
list_tuples = [('suman',12),('kiran',14),("ana",2)]
temp =  list(map(lambda x: x[1],list_tuples))
print(list(temp))
result = sorted(temp)
print(result)
```

```python
list_tuples = [('suman',12),('kiran',14),("ana",2)]
result = sorted(list_tuples, key= lambda x: x[1], reverse=False)
print(result)
```

```python
list_tuples = [('suman',12),('kiran',14),("ana",2)]
result = sorted(list_tuples, key= lambda x: x[1], reverse=True)
print(result)
```
