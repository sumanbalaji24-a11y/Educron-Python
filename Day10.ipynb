## Dictionary
1. stores a collection of key:value pairs
2. dulicate keys are not allowed
3. keys must be immutable
4. values can be anything
5. Dictionary is mutable
6. Dictionary s ordered collection

```python
# Empty Dictionary
empty_dict = {}
empty_dict1 = dict()
print(type(empty_dict),type(empty_dict1))
```

```python
## Ways to create a dictionary 
person ={"name":"suman","age":20} ## using {}
print(person)
person1 = dict(name="suman",age=20); ## using constructor
print(person)
person2 = dict([("name","suman"),("age",20)]) ## By passing List[()] list of tuples
print(person2)
```

```python
## Accessing Dictionary Items
person ={"name":"suman","age":20}
print(person["name"], person["age"])
print(person.get("name"),person.get("age"))
```

```python
## accessing nested dictionary items
person ={"name":"suman","age":20, "address":{"city":"bangalore","state":"karnataka"}}
print(person["address"]["city"])
print(person.get("address").get("city"))
```

```python
## adding the elements to dictionary
person ={"name":"suman","age":20}
person["city"] ="bangalore"
print(person)
```

```python
## Modifying existing items
person ={"name":"suman","age":20}
person["age"]=30
print(person)
```

```python
## adding multiple items
person ={"name":"suman","age":20}
person.update({"city":"Bangalore","state":"karnataka"});
print(person)
person.update({"name":"kiran","age":"45"})
print(person)
```

```python
## Reomve items 
person ={"name":"suman","age":20}
del person["name"],person["age"]
print(person)
```

```python
## Pop Removes and returns a removed value
person ={"name":"suman","age":20}
removedItem = person.pop("age")
print(person)
print(removedItem)
```

```python
## pop removes and if not found returns default
person ={"name":"suman","age":20}
removedItem = person.pop("Hello","Not Found")
print(person)
print(removedItem)
```
