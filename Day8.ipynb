##      Pop(index) removes and returns the removed element

```python
numbers = [1,2,3,4]
removedElement = numbers.pop()
print(removedElement)
print(numbers)
# numbers = []
# numbers.pop()
numbers = [1,2]
firstElement=numbers.pop(0)
print(firstElement)
print(numbers)
numbers.clear() ## Initializes to empty list
print(numbers)
```

## Sort(key, reverse)
key = function to be called before comparision(ex- lower)
reverse = true; returns desc
reverse = falase; returns asc default one.

```python
numbers = [100,300,200,700]
numbers.sort()
numbers
```

```python
numbers = [100,300,200,700]
numbers.sort()
numbers
```

```python
names = ["Apple","animal","Ball","cat"]
names.sort()
names ## words starting with Capital letter will come first
```

```python
names = ["Apple","animal","Ball","cat"]
names.sort(key=str.lower)  ## convert to lower and compare
names
```

## In and Not in returns true/false

```python
names = ["Apple","animal","Ball","cat"]
item = "animal"
print(item in names)
item = "animals"
print(item in names)
```

## Index(item,[start,end]) 
returns the first occurance of a specified item

```python
numbers = (1,2,3,4,1)
indexOf1 = numbers.index(1) ## always returns the index of first occurance of a specified item
print(indexOf1) 
```
