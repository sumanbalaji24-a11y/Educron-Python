```python
class BankAccount:
    def __init__(self):
        self.balance = 0
    def print_balance(self):
        print(self.balance)
    def deposit(self, amount):
        if(amount<0):
            print("Invalid Amount")
            return
        self.balance+= amount
    def withdraw(self, amount):
        if(amount<0 or amount > self.balance):
            print("Invalid Amount")
            return
        self.balance-= amount
```

```python
bank1 = BankAccount()
bank1.print_balance()
bank1.deposit(1000)
bank1.print_balance()
bank1.withdraw(3000)
bank1.print_balance()

```
