```python
number = 5
total= 0
while(number>=1):
    total+=number
    number-= 1
print(total)
```

```python
## Digit Counter
number=1234567
input = str(number);
totalDigits = 0
while(len(input) > 0):
    totalDigits+=1
    input  = str(number)[totalDigits:len(str(number)):1]
print(f"{totalDigits} Total Digits")
```

```python
## digit count using while loop
number = 12345678
temp = number
totalDigits = 0
while(temp>0):
    totalDigits +=1
    temp = temp // 10
print(totalDigits)
print(number)
```

```python
## find the sum of digits
number = 12346
temp = number
totalSum= 0 
while(temp>0):
    last_digit = temp %10
    totalSum+=last_digit
    temp = temp //10
print(totalSum)
```
