## fstrip, lstrip, strip

```python
text = "    123   ";
print(text.strip())
print(text.lstrip())
print(text.rstrip())
text1 = "//@aede"
print(text1.strip("/@e"))
print(text1.lstrip("/@e"))
print(text1.rstrip("/@e"))

```

## Split

```python
name ="suman balaji from kolar and i do stay in bangalore"
# by default splits by space
print(name.split()) 
print(name.split("b"))
print(name.split(' ',maxsplit=2))
```
