## Nested Loop
1. For Every outer loop you will get a inner loop executed

```python
for i in range(3):
    print(f"{i} Outer loop")
    for j in range(2):
        print(f"{j} Inner loop")
print("loop Completed")
```

```python
for i in range(1,5):
    for j in range(i):
      print(f"*", end=" ")
    print()
```

```python
for i in range(4,0,-1):
    for j in range(i):
        print("*", end=" ")
    print()

```

## List Comprehension

```python
numbers = [2,5,6,7,8,3]
result = [i*2 for i in numbers]
print(result)
```

```python
numbers = [2,5,6,7,8,3]
result = [i for i in numbers if i>5]
print(result)
print(numbers)
```

```python
numbers = [-2,5,6,7,8,-3,0]
result = [x*2 if x >0 else x for x in numbers]
print(result)
```

```python
numbers = [-1,2,3,-4]
result = ["Possitive" if number>0 else "Negative" for number in numbers]
result
```
